```python
import json
```

```python
def load_data(filename):
    with open(filename, "r") as f:
        data=json.load(f)
    return data
```

```python
data = load_data("data_introduction.json")
```

```python
def display_users(data):
    print("Users and their Connections \n")
    for user in data["users"]:
        print(f"ID : {user['id']} -> {user['name']} is friend of {user['friends']} and liked pages is/are {user['liked_pages']}")
    print("\nPages information :")
    for page in data['pages']:
        print(f"{page['id']} -> {page['name']}")

display_users(data)
```
